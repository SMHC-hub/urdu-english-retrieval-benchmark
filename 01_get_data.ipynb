```python
import os
from beir import util
from beir.datasets.data_loader import GenericDataLoader

os.makedirs("data", exist_ok=True)

paths = {}
for dataset in ["scifact", "nfcorpus"]:
    url = f"https://public.ukp.informatik.tu-darmstadt.de/thakur/BEIR/datasets/{dataset}.zip"
    paths[dataset] = util.download_and_unzip(url, "data")
    print(f"Downloaded {dataset} -> {paths[dataset]}")
```

```python
for dataset in ["scifact", "nfcorpus"]:
    corpus, queries, qrels = GenericDataLoader(data_folder=paths[dataset]).load(split="test")
    print(f"\n=== {dataset} ===")
    print("Documents in corpus :", len(corpus))
    print("Test queries        :", len(queries))
    print("Queries with labels :", len(qrels))
```
