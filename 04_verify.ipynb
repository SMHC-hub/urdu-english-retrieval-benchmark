```python
import json, random
from openpyxl import Workbook
from openpyxl.styles import Font, Alignment
from openpyxl.worksheet.datavalidation import DataValidation

random.seed(42)  # reproducible sample
SAMPLE_PER_DS = 50

def load_jsonl(path):
    with open(path, encoding="utf-8") as f:
        return [json.loads(l) for l in f]

rows = []
for ds in ["scifact", "nfcorpus"]:
    data = load_jsonl(f"queries/{ds}_cs.jsonl")
    for r in random.sample(data, min(SAMPLE_PER_DS, len(data))):
        rows.append({"id": r["id"], "dataset": ds, "EN": r["EN"],
                        "ur_en": r["ur_en"], "roman": r["roman"]})

wb = Workbook()
ws = wb.active
ws.title = "verify"
headers = ["id", "dataset", "EN", "ur_en", "roman",
            "verdict", "ur_en_fixed", "roman_fixed", "notes"]
ws.append(headers)
for c in ws[1]:
    c.font = Font(bold=True)
ws.freeze_panes = "A2"

for r in rows:
    ws.append([r["id"], r["dataset"], r["EN"], r["ur_en"], r["roman"], "", "", "", ""])

dv = DataValidation(type="list", formula1='"OK,FIX"', allow_blank=True)
ws.add_data_validation(dv)
dv.add(f"F2:F{len(rows)+1}")

widths = {"A":10,"B":10,"C":48,"D":48,"E":48,"F":9,"G":40,"H":40,"I":26}
for col, w in widths.items():
    ws.column_dimensions[col].width = w
for row in ws.iter_rows(min_row=2):
    for cell in row:
        cell.alignment = Alignment(wrap_text=True, vertical="top")

wb.save("queries/verification_sample.xlsx")
print(f"Saved verification_sample.xlsx with {len(rows)} rows ({SAMPLE_PER_DS}/dataset)")
```
